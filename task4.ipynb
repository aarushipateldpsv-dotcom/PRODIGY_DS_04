```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/twitter_training.csv")

df.head()
```

```python
df.columns = ["ID","Topic","Sentiment","Tweet"]
```

```python
df.info()
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
sns.countplot(x="Sentiment", data=df)

plt.title("Sentiment Distribution")
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(x="Topic", hue="Sentiment", data=df)

plt.xticks(rotation=90)
plt.title("Sentiment by Topic")
plt.show()
```

```python
top_topics = df["Topic"].value_counts().head(10)

plt.figure(figsize=(10,5))
top_topics.plot(kind="bar")

plt.title("Top Discussed Topics")
plt.ylabel("Number of Tweets")
plt.show()
```

```python
sentiment_counts = df["Sentiment"].value_counts()

plt.pie(sentiment_counts,
        labels=sentiment_counts.index,
        autopct='%1.1f%%')

plt.title("Overall Sentiment Ratio")
plt.show()
```
